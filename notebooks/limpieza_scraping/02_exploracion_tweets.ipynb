# Exploración de Tweets y Comentarios Ciudadanos sobre la Malla Vial de Bogotá

Este notebook permite auditar y explorar los dos datasets generados por el scraper (`src/scraping/03_descubrir_tweets.py`):
1. **`comentarios_ciudadanos.csv`**: Respuestas y quejas directas de los ciudadanos (insumo clave para percepción y análisis de sentimiento con RoBERTuito).
2. **`tweets_principales.csv`**: Publicaciones originales (denuncias principales, reportes o anuncios oficiales de UMV/IDU/Movilidad).
3. **`tweets_y_comentarios_completos.csv`**: Base consolidada y relacionada por `id_tweet_padre` e `id_conversacion`.

```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd().parents[1] if Path.cwd().name == "limpieza_scraping" else Path.cwd()
DIR_TWITTER = BASE_DIR / "data" / "twitter" / "scraping_twitter"

FILE_COMENTARIOS = DIR_TWITTER / "comentarios_ciudadanos.csv"
FILE_PRINCIPALES = DIR_TWITTER / "tweets_principales.csv"
FILE_COMPLETO = DIR_TWITTER / "tweets_y_comentarios_completos.csv"

print(f"Directorio de datos: {DIR_TWITTER}")
print(f"¿Existen comentarios?: {FILE_COMENTARIOS.exists()}")
print(f"¿Existen tweets principales?: {FILE_PRINCIPALES.exists()}")
```

```python
# Carga y comparación de publicaciones vs comentarios
if FILE_COMENTARIOS.exists() and FILE_PRINCIPALES.exists():
    df_comentarios = pd.read_csv(FILE_COMENTARIOS)
    df_principales = pd.read_csv(FILE_PRINCIPALES)
    
    print(f"Total de tweets principales: {len(df_principales):,}")
    print(f"Total de comentarios ciudadanos: {len(df_comentarios):,}")
    print("\n--- Primeros comentarios ciudadanos sobre huecos y vías ---")
    display(df_comentarios[['id_registro', 'autor_username', 'responde_a_usuario', 'favoritos', 'texto']].head(5))
else:
    print("Archivos aún no generados. Ejecuta 'src/scraping/03_descubrir_tweets.py'.")
```

```python
# Distribución de comentarios por entidad interpelada (@UMVBogota, @idubogota, etc.)
if 'df_comentarios' in locals() and not df_comentarios.empty:
    print("Comentarios por entidad o usuario al que responde:")
    display(df_comentarios['responde_a_usuario'].value_counts().head(10))
    
    print("\nComentarios con evidencia fotográfica (fotos de huecos/vías):")
    display(df_comentarios['tiene_multimedia'].value_counts())
```

```python
# Comentarios más respaldados por otros ciudadanos (mayor cantidad de likes)
if 'df_comentarios' in locals() and not df_comentarios.empty:
    top_comentarios = df_comentarios.sort_values(by='favoritos', ascending=False).head(10)
    display(top_comentarios[['autor_username', 'favoritos', 'responde_a_usuario', 'texto']])
```
